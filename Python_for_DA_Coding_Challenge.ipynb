```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('social_media_engagement_5000 (1).csv')
df
```

```python
# Check data types of all columns
print(df.dtypes)


```

```python
# Coverting into data type into datetime
df["posted_at"] = pd.to_datetime(df["posted_at"], errors="coerce")

print(df["posted_at"].head())
```

```python
# Check data types of all columns
print(df.dtypes)
```

```python
df.isnull().sum()
```

```python
df
```

```python
# imputing null values in age
df["age"].fillna(df["age"].mode(), inplace=True)
```

```python
# Droping the null value
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
# Detect duplicates
print(df.duplicated().sum())
```

```python
#Standardize categories
df["gender"] = df["gender"].str.upper().str.strip()
df["gender"] = df["gender"].replace({"M":"Male", "F":"Female"})
df
```

```python
# Correct unrealistic values in likes, comments, shares
df = df[(df["likes"] >= 0) & (df["shares"] >= 0) & (df["comments"] >= 0)]
df
```

```python
# Remove '#' from each hashtag in the list
df["hashtags"] = df["hashtags"].str.replace("#", "")
df
```

```python
# Capitalize the first letter in the 'sentiment' column
df["sentiment"] = df["sentiment"].str.capitalize()
df
```

```python
# First few rows
print('\nHead\n',df.head())

# Last few rows
print('\nTail\n',df.tail())

# Shape (rows, columns)
print('\nShapr\n',df.shape)

# Column names
print('\nColumn Name\n',df.columns)
```

```python
# Data types of each column
print('\ndtype\n',df.dtypes)

# Detailed info (non-null counts, memory usage)
print('\nInfo\n',df.info())

```

```python
# Summary stats for numeric columns
print(df.describe())
```

```python
# Value counts for a categorical column (example: post_type)
print(df["post_type"].value_counts())

# Unique values
print(df["country"].unique())

# Number of unique values
print(df["country"].nunique())
```

```python
# Correlation among numeric columns
correlation_matrix = df.corr(numeric_only=True)
print(correlation_matrix)
```

```python
# Average likes by post type
print(df.groupby("post_type")["likes"].mean())

# Total impressions by country
print(df.groupby("country")["impression_count"].sum())
```

```python
# Create new fields such as engagement_score and hashtag count.

df["engagement_score"] = df["likes"] + df["shares"] + df["comments"]
df["hashtag_count"] = df["hashtags"].str.split().str.len().fillna(0).astype(int)

df
```

```python
#groupby Post Type:
print(df.groupby("post_type")["engagement_score"].mean())
#groupby country Type:
print(df.groupby("country")["engagement_score"].sum())
#groupby sentiment Type:
print(df.groupby("sentiment")["engagement_score"].mean())
```

```python
cols = ["likes", "comments", "shares", "watch_time_sec", "engagement_rate", "follower_count"]
data = df[cols]

# Mean
print('\nMean:\n',data.mean())

# Median
print('\nMedian:\n',data.median())

# Mode (can return multiple values if multimodal)
print('\nMode:\n',data.mode().iloc[0])
```

```python
# Line Plot: Daily Engagement Trend
daily_engagement = df.groupby("posted_at")["engagement_score"].sum()

plt.plot(daily_engagement.index, daily_engagement.values, marker="o",color='slategray')
plt.xlabel("Date")
plt.ylabel("Engagement Score")
plt.title("Daily Engagement Trend")
plt.grid(axis='y',linestyle=':',color='black')
plt.xticks(rotation=45)
```

```python
# Bar Chart: Posts by Category
post_type_counts = df["post_type"].value_counts()
post_type_counts.plot(kind="bar",color='gray')
plt.xlabel("Post Type")
plt.ylabel("Count")
plt.title("Posts by Category")
plt.grid(axis='y',linestyle=':',color='black')
# Loop through the aggregated data to add value labels
for i, (post_type, count) in enumerate(post_type_counts.items()):
    plt.text(i, count, f'{count}', ha='center', va='bottom')
```

```python
#Pie Chart: Gender Distribution
df["gender"].value_counts().plot(kind="pie", autopct="%1.1f%%",colors=['dimgrey','gray','darkgray'])
plt.title("Gender Distribution")
plt.ylabel("")

```

```python
# Histogram: Age
plt.hist(df["age"], bins=10,color='dimgray', edgecolor="black")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.grid(axis='y',linestyle=':',color='black')
```

```python
# Box Plot: Engagement Rate
plt.boxplot(df["engagement_rate"].dropna())
plt.ylabel("Engagement Rate")
plt.title("Box Plot: Engagement Rate")
plt.grid(axis='y',linestyle=':',color='black')
```

```python
#Count Plot: Post Type
sns.countplot(x="post_type", data=df, palette=['dimgrey','gray','darkgray','silver'])
plt.title("Count Plot: Post Type")
plt.grid(axis='y',linestyle=':',color='black')
for i, (post_type, count) in enumerate(post_type_counts.items()):
    plt.text(i, count, f'{count}', ha='center', va='bottom')
```

```python
sns.barplot(x="post_type", y="likes", data=df, estimator="mean", palette=['dimgrey','gray','darkgray','silver'])
plt.title("Average Likes by Post Type")
plt.grid(axis='y',linestyle=':',color='black')
post_type_mean = df.groupby("post_type")["likes"].mean()
for i, (post_type, mean) in enumerate(post_type_mean.items()):
    plt.text(i, mean, f'{mean:.0f}', ha='center', va='bottom')
```

```python
# Violin Plot: Followers vs Sentiment
sns.violinplot(x="sentiment", y="follower_count", data=df, palette=['dimgrey','gray','darkgray','silver'])
plt.title("Followers Distribution by Sentiment")
plt.grid(axis='y',linestyle=':',color='black')
```

```python
# Heatmap: Correlation Matrix
numeric_cols = df.select_dtypes(include=['number']).columns
corr = df[numeric_cols].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix Heatmap")
```
